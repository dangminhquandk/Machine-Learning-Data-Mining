3. Phân chia dữ liệu train - test

```python
# Xác định thông tin thuộc tính X và nhãn y
X = df.drop("price", axis=1)
y = df['price']
print('Shape of X = ', X.shape)
print('Shape of y = ', y.shape)
```

```python
#Chia dữ liệu - train_test_split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 51)
print('Shape of X_train = ', X_train.shape)
print('Shape of y_train = ', y_train.shape)
print('Shape of X_test = ', X_test.shape)
print('Shape of y_test = ', y_test.shape)
```

4. Feature Scaling

> If feature scaling is not done, then a machine learning algorithm tends to weigh greater values, higher and consider smaller values as the lower values, regardless of the unit of the values.

```python
#Chuẩn hóa giá trị của các feature trong 1 phạm vi nào đó.

from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
sc.fit(X_train)
X_train= sc.transform(X_train)
X_test = sc.transform(X_test)
```

## Xây dựng mô hình huấn luyện

### Linear Regression

```python
from sklearn.linear_model import LinearRegression
from sklearn.linear_model import Lasso
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error
lr = LinearRegression()
lr_lasso = Lasso()
lr_ridge = Ridge()
```

```python
def rmse(y_test, y_pred):
  return np.sqrt(mean_squared_error(y_test, y_pred))
```

```python
lr.fit(X_train, y_train)
lr_score = lr.score(X_test, y_test) # with all num var 0.7842744111909903
lr_rmse = rmse(y_test, lr.predict(X_test))
lr_score, lr_rmse
```

```python
# Lasso
lr_lasso.fit(X_train, y_train)
lr_lasso_score=lr_lasso.score(X_test, y_test) # with balcony 0.5162364637824872
lr_lasso_rmse = rmse(y_test, lr_lasso.predict(X_test))
lr_lasso_score, lr_lasso_rmse
```

```python
# Ridge
lr_ridge.fit(X_train, y_train)
lr_ridge_score = lr_ridge.score(X_test, y_test) # with all num var 0.7842744111909903
lr_ridge_rmse = rmse(y_test, lr_ridge.predict(X_test))
lr_ridge_score, lr_ridge_rmse
```

### Support Vector Machine

```python
from sklearn.svm import SVR
svr = SVR()
svr.fit(X_train,y_train)
svr_score=svr.score(X_test,y_test)
svr_rmse = rmse(y_test, svr.predict(X_test))
svr_score, svr_rmse
```

### Random Forest

```python
from sklearn.ensemble import RandomForestRegressor
rfr = RandomForestRegressor()
rfr.fit(X_train,y_train)
rfr_score=rfr.score(X_test,y_test)
rfr_rmse = rmse(y_test, rfr.predict(X_test))
rfr_score, rfr_rmse
```

## Test Model

```python
def predict_house_price(model, bath, balcony, total_sqft_float, bhk, area_type, availability, location):
    x = np.zeros(len(X.columns))
    x[0] = bath
    x[1] = balcony
    x[2] = total_sqft_float
    x[3] = bhk

    if "availability_Ready To Move" in X.columns and availability == "Ready To Move":
        x[X.columns.get_loc("availability_Ready To Move")] = 1

    if "area_type_" + area_type in X.columns:
        area_type_index = np.where(X.columns == "area_type_" + area_type)[0][0]
        x[area_type_index] = 1

    if "location_" + location in X.columns:
        loc_index = np.where(X.columns == "location_" + location)[0][0]
        x[loc_index] = 1

    x = sc.transform([x])[0]
    return model.predict([x])[0]
```

## 2.5 Bảng kết quả so sánh tổng hợp (Bảng KQua)
Dưới đây là bảng tổng hợp so sánh các trọng số ($w$), hệ số chặn ($Intercept$) và sai số kiểm thử **RMSE** của 3 mô hình tuyến tính: **OLS (Ordinary Least Squares)**, **Ridge**, và **Lasso** sau khi đã loại bỏ trường `price_per_sqft` (tránh rò rỉ dữ liệu).

```python
# ===============================================================
# BẢNG KẾT QUẢ TỔNG HỢP (BẢNG KQua)
# So sánh các mô hình: OLS, Ridge, Lasso
# Đã loại bỏ 'price_per_sqft' (tránh rò rỉ dữ liệu / Target Leakage)
# Đổi thước đo sai số RSS_Test thành RMSE theo yêu cầu
# ===============================================================

# 1. Dự đoán trên tập kiểm thử và tính RMSE (Root Mean Squared Error)
y_pred_ols = lr.predict(X_test)
y_pred_ridge = lr_ridge.predict(X_test)
y_pred_lasso = lr_lasso.predict(X_test)

rmse_ols = rmse(y_test, y_pred_ols)
rmse_ridge = rmse(y_test, y_pred_ridge)
rmse_lasso = rmse(y_test, y_pred_lasso)

# 2. Xây dựng Bảng KQua
records = []
for col, w_lr, w_ridge, w_lasso in zip(X.columns, lr.coef_, lr_ridge.coef_, lr_lasso.coef_):
    records.append({
        'Thuộc tính': col,
        'OLS': round(w_lr, 4),
        'Ridge': round(w_ridge, 4),
        'Lasso': round(w_lasso, 4)
    })

# Dòng Intercept
records.append({
    'Thuộc tính': 'Intercept',
    'OLS': round(lr.intercept_, 4),
    'Ridge': round(lr_ridge.intercept_, 4),
    'Lasso': round(lr_lasso.intercept_, 4)
})

# Dòng cuối: RMSE (Thay cho RSS_Test)
records.append({
    'Thuộc tính': 'RMSE',
    'OLS': round(rmse_ols, 4),
    'Ridge': round(rmse_ridge, 4),
    'Lasso': round(rmse_lasso, 4)
})

df_kqua = pd.DataFrame(records)
print("=== Bảng KQua (OLS vs Ridge vs Lasso) ===")
display(df_kqua)

# 3. In phân tích nhận xét chuyên sâu
print("\n[GIẢI THÍCH & ĐÁNH GIÁ CHUYÊN SÂU]:")
print("1. Loại bỏ thuộc tính 'price_per_sqft':")
print("   - Cột 'price_per_sqft' vốn được tính từ price / total_sqft. Nếu đưa cột này vào làm biến độc lập để dự đoán 'price', mô hình sẽ bị rò rỉ nhãn (Target Leakage) và mất tính ứng dụng thực tế.")
print("   - Khi loại bỏ 'price_per_sqft', mô hình phải học mối quan hệ thực sự giữa diện tích (total_sqft), số phòng và đặc biệt là VỊ TRÍ (location) để định giá nhà.")
print("2. Đổi RSS_Test thành RMSE:")
print("   - RMSE có cùng đơn vị với giá nhà (Lakhs Rupee), đo lường độ lệch trung bình thực tế trên mỗi căn nhà thay vì con số tổng bình phương (Lakhs^2) khổng lồ của RSS.")
print("3. So sánh các mô hình tuyến tính:")
print(f"   - OLS: RMSE bùng nổ (~{rmse_ols:.2e}) do hiện tượng đa cộng tuyến giữa hàng trăm biến One-Hot Encoding.")
print(f"   - Ridge (alpha=1.0): RMSE = {rmse_ridge:.4f} Lakhs -> Kiểm soát đa cộng tuyến tốt.")
print(f"   - Lasso (alpha=0.1): RMSE = {rmse_lasso:.4f} Lakhs -> Tối ưu nhất (RMSE thấp nhất) và triệt tiêu bớt trọng số nhiễu về 0.")
```
